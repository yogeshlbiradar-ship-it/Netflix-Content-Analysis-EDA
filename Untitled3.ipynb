```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Prepare data
content_data = df_clean['type'].value_counts()

rating_data = (
    df_clean['rating']
    .value_counts()
    .head(8)
)

genre_data = (
    df_clean['listed_in']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

year_data = (
    df_clean['release_year']
    .value_counts()
    .sort_index()
)

# Create dashboard layout
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Movies vs TV Shows',
        'Top 8 Content Ratings',
        'Top 10 Genres',
        'Content by Release Year'
    ),
    specs=[
        [{'type': 'domain'}, {'type': 'xy'}],
        [{'type': 'xy'}, {'type': 'xy'}]
    ]
)

# Chart 1: Movies vs TV Shows
fig.add_trace(
    go.Pie(
        labels=content_data.index,
        values=content_data.values,
        hole=0.4
    ),
    row=1,
    col=1
)

# Chart 2: Content Ratings
fig.add_trace(
    go.Bar(
        x=rating_data.index,
        y=rating_data.values,
        name='Ratings'
    ),
    row=1,
    col=2
)

# Chart 3: Top Genres
fig.add_trace(
    go.Bar(
        x=genre_data.values,
        y=genre_data.index,
        orientation='h',
        name='Genres'
    ),
    row=2,
    col=1
)

# Chart 4: Release Year Trend
fig.add_trace(
    go.Scatter(
        x=year_data.index,
        y=year_data.values,
        mode='lines+markers',
        name='Release Trend'
    ),
    row=2,
    col=2
)

# Dashboard formatting
fig.update_layout(
    title='Netflix Interactive Overview Dashboard',
    height=800,
    width=1100,
    showlegend=True
)

fig.update_xaxes(title_text='Rating', row=1, col=2)
fig.update_yaxes(title_text='Number of Titles', row=1, col=2)

fig.update_xaxes(title_text='Number of Titles', row=2, col=1)
fig.update_xaxes(title_text='Release Year', row=2, col=2)
fig.update_yaxes(title_text='Number of Titles', row=2, col=2)

fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Movie duration
movie_df = df_clean[df_clean['type'] == 'Movie'].copy()

movie_df['duration_minutes'] = (
    movie_df['duration']
    .str.extract(r'(\d+)')
    .astype(float)
)

# TV Show seasons
tv_df = df_clean[df_clean['type'] == 'TV Show'].copy()

tv_df['seasons'] = (
    tv_df['duration']
    .str.extract(r'(\d+)')
    .astype(float)
)

# Top 10 directors
director_df = df_clean.dropna(subset=['director']).copy()

director_df['director'] = director_df['director'].str.split(',')

director_df = director_df.explode('director')

director_df['director'] = director_df['director'].str.strip()

director_data = director_df['director'].value_counts().head(10)

# Create dashboard
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Movie Duration Distribution',
        'TV Shows by Number of Seasons',
        'Top 10 Directors',
        'Movie Duration Box Plot'
    )
)

# Chart 1: Movie duration histogram
fig.add_trace(
    go.Histogram(
        x=movie_df['duration_minutes'],
        name='Movie Duration'
    ),
    row=1,
    col=1
)

# Chart 2: TV seasons
season_data = tv_df['seasons'].value_counts().sort_index()

fig.add_trace(
    go.Bar(
        x=season_data.index,
        y=season_data.values,
        name='TV Seasons'
    ),
    row=1,
    col=2
)

# Chart 3: Directors
fig.add_trace(
    go.Bar(
        x=director_data.values,
        y=director_data.index,
        orientation='h',
        name='Directors'
    ),
    row=2,
    col=1
)

# Chart 4: Box plot
fig.add_trace(
    go.Box(
        y=movie_df['duration_minutes'],
        name='Movie Duration'
    ),
    row=2,
    col=2
)

fig.update_layout(
    title='Netflix Content Analysis Dashboard',
    height=850,
    width=1200,
    showlegend=False
)

fig.update_xaxes(title_text='Duration (Minutes)', row=1, col=1)
fig.update_yaxes(title_text='Number of Movies', row=1, col=1)

fig.update_xaxes(title_text='Number of Seasons', row=1, col=2)
fig.update_yaxes(title_text='Number of TV Shows', row=1, col=2)

fig.update_xaxes(title_text='Number of Titles', row=2, col=1)
fig.update_xaxes(title_text='Duration (Minutes)', row=2, col=2)

fig.show()
```

```python
# Netflix Content Analysis — Final Analytical Report

## 1. Introduction

Netflix is one of the popular streaming platforms offering Movies and TV Shows across different countries. Analyzing its content catalog can help us understand the types of content available, genres, ratings, countries, release years, and content addition trends.

This project performs Exploratory Data Analysis (EDA) on a Netflix dataset using Python. The analysis uses Pandas for data processing and visualization libraries such as Matplotlib and Plotly to create charts and dashboards.

## 2. Project Objectives

The main objectives of this project are:

- To perform complete Exploratory Data Analysis on the Netflix dataset.
- To identify key content trends and patterns.
- To analyze Movies and TV Shows based on type, rating, genre, country, and release year.
- To develop multiple visual dashboards.
- To generate actionable business insights.
- To prepare a final analytical report based on the findings.

## 3. Dataset Description

The Netflix dataset contains information about titles available on the platform.

The dataset includes the following columns:

| Column | Description |
|---|---|
| `show_id` | Unique identifier for each title |
| `type` | Type of content: Movie or TV Show |
| `title` | Name of the content |
| `director` | Director of the content |
| `country` | Country where the content was produced |
| `date_added` | Date when the content was added to Netflix |
| `release_year` | Year the content was released |
| `rating` | Content rating |
| `duration` | Movie duration or number of TV Show seasons |
| `listed_in` | Genre or category of the content |

## 4. Tools and Technologies

The following tools and libraries were used:

- **Python:** Programming language used for analysis.
- **Jupyter Notebook:** Environment for executing code and documenting results.
- **Pandas:** Data cleaning, manipulation, and analysis.
- **Matplotlib:** Data visualization.
- **Plotly:** Interactive visualizations and dashboards.
- **NumPy:** Numerical operations.

## 5. Data Cleaning and Preprocessing

Before performing the analysis, the dataset was cleaned and prepared.

The following preprocessing steps were performed:

1. Created a separate DataFrame for analysis.
2. Removed unnecessary spaces from text columns.
3. Standardized values in the content type and rating columns.
4. Converted the `date_added` column into datetime format.
5. Checked missing values in the dataset.
6. Checked duplicate records.
7. Extracted year and month from the `date_added` column.
8. Prepared separate data for country and genre analysis.

These steps helped improve data quality and made the dataset suitable for analysis.

## 6. Exploratory Data Analysis (EDA)

Exploratory Data Analysis was performed to understand the structure and characteristics of the Netflix content catalog.

### 6.1 Content Type Analysis

The dataset was analyzed to compare Movies and TV Shows.

A pie chart was created to visualize the proportion of each content type. This analysis helps understand the composition of Netflix's content library.

### 6.2 Rating Analysis

The distribution of content ratings was analyzed using a bar chart.

This helps identify which rating categories are common in the dataset and understand the classification of content available on Netflix.

### 6.3 Genre Analysis

The `listed_in` column was used to analyze the genres and categories of content.

Genre values were separated to study the frequency of different categories. A bar chart was created to display the most frequently listed genres.

### 6.4 Release Year Analysis

The `release_year` column was analyzed to understand the distribution of content across different years.

A trend chart was created to observe how the release years of Netflix titles are distributed.

### 6.5 Content Addition Trend

The `date_added` column was used to analyze when titles were added to Netflix.

The year and month were extracted from the date. Visualizations were created to explore content additions over time and identify seasonal patterns.

### 6.6 Country Analysis

The `country` column was analyzed to identify countries contributing content to Netflix.

For titles associated with multiple countries, the country values were separated for analysis. A chart was created to display countries with a high number of titles.

### 6.7 Movie Duration Analysis

The duration of Movies was analyzed using a histogram and box plot.

This helps understand the distribution of movie durations, identify common duration ranges, and observe possible outliers.

### 6.8 TV Show Season Analysis

The duration column was used to analyze the number of seasons in TV Shows.

This analysis helps understand how TV Show seasons are distributed across the catalog.

### 6.9 Director Analysis

The director column was analyzed to identify directors associated with multiple Netflix titles.

A bar chart was created to display directors with a high number of titles.

## 7. Visual Dashboards

Three visual dashboards were developed to present the analysis in a clear and understandable way.

### Dashboard 1 — Netflix Overview

This dashboard includes:

- Distribution of Movies and TV Shows
- Content rating distribution
- Top genres
- Release-year trends

**Purpose:** To provide an overview of the Netflix content catalog.

### Dashboard 2 — Content Trends

This dashboard includes:

- Content additions by year
- Content additions by month
- Top countries contributing content
- Country-wise content type analysis

**Purpose:** To understand the growth and distribution of Netflix content over time and across countries.

### Dashboard 3 — Content Analysis

This dashboard includes:

- Movie duration distribution
- TV Show season analysis
- Top directors
- Movie duration box plot

**Purpose:** To explore content characteristics in greater detail.

## 8. Key Business Insights

The following insights can be investigated using the results of the analysis:

1. **Content Mix:** The comparison of Movies and TV Shows helps understand the composition of Netflix's catalog.
2. **Rating Distribution:** Rating categories provide information about the classification of available content.
3. **Genre Trends:** Genre analysis helps identify categories that are frequently represented.
4. **Content Growth:** Content addition trends show how the catalog has changed over time.
5. **Regional Contribution:** Country analysis helps identify major content-producing regions.
6. **Movie Duration:** Duration analysis helps understand common movie length ranges.
7. **TV Show Structure:** Season analysis provides an overview of the structure of TV Shows.
8. **Director Contribution:** Director analysis identifies individuals associated with multiple titles.

The exact values and rankings should be taken from the outputs and charts generated in the notebook.

## 9. Actionable Business Recommendations

Based on the analysis, the following recommendations can be considered:

### 9.1 Content Planning

Netflix can use genre and content type analysis to review its catalog composition and identify categories that may require further investigation.

### 9.2 Regional Content Strategy

Country-wise analysis can help Netflix explore regional content opportunities and understand the contribution of different countries.

### 9.3 Catalog Management

Release-year and content addition trends can help in reviewing the age and growth of the content catalog.

### 9.4 Content Duration

Movie duration and TV Show season analysis can support catalog categorization and help understand the range of content formats available.

### 9.5 Further Research

The analysis can be extended by combining the dataset with audience viewing data, ratings, watch time, or revenue information to better evaluate content performance.

## 10. Limitations

This project has certain limitations:

- The dataset contains catalog information but does not provide viewer counts or watch time.
- Revenue and subscription data are not available.
- The presence of a title in the dataset does not indicate its popularity.
- Country and genre fields may contain multiple values for a single title.
- Business recommendations are based on catalog patterns and should be validated using additional business data.

## 11. Conclusion

This project performed Exploratory Data Analysis on the Netflix dataset using Python and data visualization tools.

The analysis explored content types, ratings, genres, release years, content addition trends, countries, movie durations, TV Show seasons, and directors.

Three visual dashboards were developed to communicate the findings. The project demonstrates how data cleaning, exploratory analysis, and visualization can be used to understand a real-world dataset and support data-driven decision-making.

Overall, this project provides a foundation for further analysis of Netflix's content catalog and business strategy.

## 12. Future Scope

The project can be improved in the future by:

- Adding interactive filters to dashboards.
- Integrating viewer ratings and audience engagement data.
- Comparing content trends across countries.
- Performing time-series analysis on content additions.
- Using machine learning to explore content classification or recommendation systems.

## 13. References

- Netflix Dataset
- Pandas Documentation
- Matplotlib Documentation
- Plotly Documentation
- Python Documentation
```
