```python
import plotly.express as px
import pandas as pd

# 1. Movies vs TV Shows
content_data = (
    df['type']
    .value_counts()
    .rename_axis('type')
    .reset_index(name='count')
)

fig1 = px.pie(
    content_data,
    names='type',
    values='count',
    title='Movies vs TV Shows Distribution',
    hole=0.4
)
fig1.show()


# 2. Top 10 Genres
genre = df['listed_in'].dropna().str.split(',').explode().str.strip()

genre_data = (
    genre.value_counts()
    .head(10)
    .rename_axis('genre')
    .reset_index(name='count')
)

fig2 = px.bar(
    genre_data,
    x='count',
    y='genre',
    orientation='h',
    title='Top 10 Netflix Genres'
)
fig2.show()


# 3. Content Ratings
rating_data = (
    df.groupby(['type', 'rating'])
    .size()
    .reset_index(name='count')
)

fig3 = px.bar(
    rating_data,
    x='rating',
    y='count',
    color='type',
    barmode='group',
    title='Ratings Across Movies and TV Shows'
)
fig3.show()


# 4. Content by Release Year
year_data = (
    df['release_year'].value_counts()
    .sort_index()
    .rename_axis('release_year')
    .reset_index(name='count')
)

fig4 = px.line(
    year_data,
    x='release_year',
    y='count',
    markers=True,
    title='Netflix Content by Release Year'
)
fig4.show()
```

```python
import pandas as pd

# 1. Movies vs TV Shows
content_type = df['type'].value_counts()
content_percent = df['type'].value_counts(normalize=True) * 100

print("1. CONTENT TYPE INSIGHTS")
for name, count in content_type.items():
    print(f"{name}: {count} titles ({content_percent[name]:.2f}%)")


# 2. Most common content ratings
print("\n2. MOST COMMON CONTENT RATINGS")
print(df['rating'].value_counts().head(5))


# 3. Top 5 genres
genre = df['listed_in'].dropna().str.split(',').explode().str.strip()

print("\n3. TOP 5 GENRES")
print(genre.value_counts().head(5))


# 4. Movies and TV Shows by rating
print("\n4. RATINGS BY CONTENT TYPE")
print(pd.crosstab(df['type'], df['rating']))


# 5. Content released in recent years
print("\n5. RECENT RELEASE TRENDS")
print(df['release_year'].value_counts().sort_index().tail(10))
```
